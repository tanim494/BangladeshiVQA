# Explore BangladeshiVQA

A minimal viewer over the **bundled sample images** (`data/sample_images/`, 3 per
category) and their question–answer pairs. No download required.

The **full dataset** (all 2,068 images + 7,038 QA pairs + splits) lives on Hugging Face —
see the README "Download" section. The last cell shows how to pull the full annotations.


```python
import json, os
from collections import Counter
from PIL import Image
import matplotlib.pyplot as plt

REPO_ROOT = os.path.dirname(os.getcwd()) if os.path.basename(os.getcwd()) == 'examples' else os.getcwd()
SAMPLE_DIR = os.path.join(REPO_ROOT, 'data', 'sample_images')
samples = json.load(open(os.path.join(SAMPLE_DIR, 'sample_qa.json'), encoding='utf-8'))
print(len(samples), 'sample images across', len({s['category'] for s in samples}), 'categories')
```

## Category distribution of the samples


```python
Counter(s['category'] for s in samples)
```

## View images with their QA pairs

Bangla text renders best with a Bangla-capable font installed. Each image is shown
with its question type, cognitive level, question, and answer.


```python
def show(sample):
    img_path = os.path.join(SAMPLE_DIR, sample['image_id'] + '.jpg')
    plt.figure(figsize=(5, 5))
    plt.imshow(Image.open(img_path)); plt.axis('off')
    plt.title(f"{sample['image_id']}  ·  {sample['category']}")
    plt.show()
    for qa in sample['qa_pairs']:
        print(f"[{qa['q_type']} / {qa['level']}]")
        print('  Q:', qa['question'])
        print('  A:', qa['answer'])
    print('-' * 60)

# Show the first image from each category
seen = set()
for s in samples:
    if s['category'] not in seen:
        seen.add(s['category'])
        show(s)
```

## Load the full annotations from Hugging Face (optional)

The full 7,038 QA pairs and splits are hosted on Hugging Face, not in this repo. Set the
`HF_DATASET_REPO` environment variable (or edit it below) to your dataset repo id.


```python
from huggingface_hub import hf_hub_download

HF_DATASET_REPO = os.environ.get('HF_DATASET_REPO', 'tanim494/BangladeshiVQA')  # e.g. 'your-username/BangladeshiVQA'
if '[' in HF_DATASET_REPO:
    print('Set HF_DATASET_REPO to your Hugging Face dataset repo id to load the full annotations.')
else:
    path = hf_hub_download(repo_id=HF_DATASET_REPO, filename='annotations/qa_pairs.json', repo_type='dataset')
    qa = json.load(open(path, encoding='utf-8'))
    print('Total QA pairs:', len(qa))
    print('Question types:', dict(Counter(r['q_type'] for r in qa)))
    print('Levels:', dict(Counter(r['level'] for r in qa)))
```
